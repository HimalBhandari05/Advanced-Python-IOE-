```python
# iterators 

aList = [1,2,4,5,6]
iterator = iter(aList)

while True:
    try:
        num = next(iterator)
        print(num)
    except StopIteration:
        break
```

```python
# Custom class for iterators .


class CustomCounter:
    def __init__(self, limit):
        self.limit = limit
        self.current = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < self.limit:
            self.current += 1
            return self.current
        else:
            raise StopIteration


obj = CustomCounter(12)
print(obj)
for num in obj:
    print(num, end=" ")
```

```python
def cubes(n):
    for i in range (n):
        yield i**3

for val in cubes(6):
    print(val)
    
```

```python
# Custom class for iterators .


class EvenCounter:
    def __init__(self, limit):
        self.limit = limit
        self.current = 0
        self.count = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < self.limit:
            self.current += 2
            self.count += 1
            return self.current
        else:
            raise StopIteration



obj = EvenCounter(12)
print(obj)
for num in obj:
    print(num, end=" ")
```
