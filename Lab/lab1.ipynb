```python
# 1.Write a Python program to check if a given number is prime.


def checkPrime(n):
    count = 0
    for i in range(1, n + 1):
        if n % i == 0:
            count += 1
    if count == 2:
        print(f"{n} is a prime number")
    else:
        print(f"{n} is not a prime number")


num = int(input("Enter the number to check"))
checkPrime(num)
```

```python
# 2.Create a function that converts a list of temperatures from Celsius to Fahrenheit.

list_of_temp = [100, 90, 26, 34]


def celcius_to_fahrenheit(celcius_list):
    farh_list = []
    for celcius in celcius_list:
        farh = (celcius * 1.8) + 32
        farh_list.append(farh)

    return farh_list


to_fah = celcius_to_fahrenheit(list_of_temp)


for index, value in enumerate(to_fah):
    print(f"{list_of_temp[index]} degree = {value}*F ")
```

```python
# 3.Write a program to count the number of vowels in a given string


def count_vowels(strValue):
    vowels = ["a", "e", "i", "o", "u"]
    updated_string = strValue.lower()
    final_list = []

    for i in updated_string:
        if i in vowels:
            final_list.append(i)

    vowel_count = len(final_list)
    return vowel_count


given_string = "HimalBhandari"
count = count_vowels(given_string)

print(f"The count of {given_string} is: {count}")
```

```python
# 4. Implement a function to find the factorial of a number using recursion.


def find_fact(n):
    if n < 0:
        return None
    if n == 0 or n == 1:
        return 1
    else:
        return n * find_fact(n - 1)


num = int(input("Enter the number: "))
fact_value = find_fact(num)
print(f"The factorial of {num} is {fact_value}")
```

```python
# 7. Write a program to check if a string is a palindrome.


def check_palindrome(str_value):
    updated_str = str_value.lower()
    reverse_value = updated_str[::-1]
    if updated_str == reverse_value:
        return True
    return False


given_string = "manam"
true_or_false = check_palindrome(given_string)

if true_or_false:
    print("Palindrome")
else:
    print("Not palindrome")
```

```python
# 8. Implement a class Circle with a method to calculate the area and circumference.

PI = 3.14
class Circle:
    def __init__(self, radius):
        self.radius = radius
    
    
    def area(self):
        return float(PI*(self.radius**2))
    
    def circumference(self):
        return float(2*PI*self.radius)
    


circle1 = Circle(3)
area_hai = circle1.area()
ciru_hai = circle1.circumference()


print(f"THe area and cirumference is : {area_hai} and {ciru_hai}")
```

```python
# 11. Implement a function to remove duplicates from a list without using set().

list_value = ["child", "is", "good", "good", "nice"]


def remove_duplicated(list_val):
    result = []
    for list_item in list_val:
        if list_item not in result:
            result.append(list_item)

    return result


result_list = remove_duplicated(list_value)
print(result_list)
```

```python
# 13. Create a class Rectangle with methods to calculate area and perimeter.


PI = 3.14


class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return float(self.length * self.breadth)

    def perimeter(self):
        return float(2 * (self.length + self.breadth))


rec1 = Rectangle(3,5)
area_hai = rec1.area()
per_hai = rec1.perimeter()


print(f"THe area and cirumference is : {area_hai} and {per_hai}")
```

```python
# 15. Implement a Python function that generates the Fibonacci sequence up to n terms


def fib_sequence(n):
    if n == 0:
        return []
    if n == 1:
        return [0]
    if n == 2:
        return [0, 1]

    seq = fib_sequence(n - 1)
    seq.append(seq[-1] + seq[-2])
    return seq


fibo_num = fib_sequence(5)
print(fibo_num)
```

```python
# 16. Write a program to count the frequency of each word in a given string.


def count_frequency(s):
    freq = {}

    for ch in s:
        if ch in freq:
            freq[ch] += 1
        else:
            freq[ch] = 1

    return freq


print(count_frequency("himalbhandari"))
```

```python
# 17. Create a function that accepts a variable number of arguments and returns their sum.


def sum_numbers(*args):
    total = 0
    for num in args:
        total += num
    return total


print(sum_numbers(10, 20, 30))
print(sum_numbers(5, 15, 25, 50))
```

```python
# 18. Implement a class Employee with attributes name and salary and a method to
# apply a 10% raise.


class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def apply_raise(self):
        self.salary = self.salary + (self.salary * 0.10)


emp1 = Employee("Himal", 50000)
emp1.apply_raise()
print(emp1.name, emp1.salary)
```

```python
# 19. Write a Python program to check if two strings are anagrams of each other.

def check_anagram(s1, s2):
    return sorted(s1) == sorted(s2)

print(check_anagram("listen", "silent"))

print(check_anagram("hello", "world"))
```

```python
# 20. Create a function that calculates the sum of digits of a given number.


def sum_of_digits(num):
    total = 0
    for digit in str(num):
        total += int(digit)
    return total


print(sum_of_digits(1234))
print(sum_of_digits(908))
```
