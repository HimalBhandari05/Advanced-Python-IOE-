```python
# Objective: Read an Excel file, process the data, and write the results to a new Excel file.

from openpyxl import load_workbook
import pandas as pd

workbook = load_workbook("data.xlsx")
# print(workbook.sheetnames)

sheet = workbook.active

    
# value = sheet["A1"].value
# value2 = sheet["C1"].value
# print(value2)
# print(value)


actualData = []

for row in sheet.iter_rows(values_only=True):
    actualData.append(row)
    

header = actualData[0]
# print(header)   
data = actualData[1:]
# print(data)

df = pd.DataFrame(data, columns=header)
print(df)
```

```python
# Total and average marks for each students.

df["Total"] = df['Math'] + df["Science"] + df["English"]
# print(df)

df['Average'] = df['Total']/3
print(df)
```

```python
print(df.dtypes)

df['English'] = pd.to_numeric(df['English'])
```

```python
print(df.dtypes)
```

```python
results = []

for row in df['Average']:
    if row >= 60:
        results.append("Passed")
    else:
        results.append("Failed")
        
df['Result'] = results
```

```python
df.to_excel('output.xlsx',index=False)
```

```python
# Relational database model.( rows and column ma store garxa yesle)

import sqlite3

# Connect to the database (if not then create it)


conn = sqlite3.connect("example.db")
# print(type(conn))
# print(type(cursor))
cursor = conn.cursor()


# Database connected now we should create some table.


cursor.execute(
    """
        CREATE TABLE IF NOT EXISTS student(
            id INTEGER PRIMARY KEY,
            name TEXT,
            age INTEGER
        )
    """
)

cursor.execute(
    """
        INSERT INTO student (name,age) VALUES (?,?)
        """,
    ("Himal Bhandari", 24),
)

cursor.execute(
    """
        INSERT INTO student (name,age) VALUES (?,?)
    """,
    ("Sajina Magar", 22),
)

conn.commit()  # yo chai save garejasto ho hai.
```

```python
cursor.execute("SELECT * FROM student")

rows = cursor.fetchall()
# print(rows)
for row in rows:
    print(row)

conn.close()
```

```python
# Using sql alchemy for rdbms operations.

from sqlalchemy import create_engine

engine = create_engine('sqlite:///example.db',echo=True)
# print(type(engine))
```

```python
from sqlalchemy.ext.declarative import declarative_base
from sqlalchemy import Column, Integer, String

Base = declarative_base()
# print(type(Base))


class Teacher(Base):
    __tablename__ = "teachers"
    id = Column(Integer, primary_key=True)
    name = Column(String)
    age = Column(Integer)


# now we create a table.

Base.metadata.create_all(engine)
```

```python
from sqlalchemy.orm import sessionmaker

Session = sessionmaker(bind=engine)
session = Session()
print(session)
```

```python
teacher1 = Teacher(name="Himal", age=30)

session.add(teacher1)
session.commit()
```

```python
teacher2 = Teacher(name="Sajina",age=40)
session.add(teacher2)
session.commit()
```
