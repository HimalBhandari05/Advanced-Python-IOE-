```python
# Generator -> iterator ready parxa. with yield function.
def square(n):
    for i in range(n):
        yield i**2


print(square(5))

for i in square(6):
    print(i, end=" ")
```

```python
# decorators -> function wrapper that adds the extra functionality to the existing function.


def my_decorator(func):
    def wrapper():
        print("Start")
        func()
        print("End")

    return wrapper


def middle():
    print("Middle")


middle_man = my_decorator(middle)
middle_man()
type(middle_man)
```

```python
import time


def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, *kwargs)
        end = time.time()

        return end - start

    return wrapper


@timer
def fibo(n):
    seq = []
    seq.append(0)
    a, b = 0, 1
    for i in range(n):
        c = a + b
        a = b
        b = c
        seq.append(c)

    print(seq)


fibo(34)
```

```python
students = [
    {"age": 23, "date": 2020},
    {"age": 35, "date": 6320},
    {"age": 22, "date": 5020},
    {"age": 33, "date": 2320},
    {"age": 63, "date": 5024},
]

sorted_values = sorted(students, key=lambda x: x["date"])
print(sorted_values)
```

```python
# filtering .

students = [
    {"age": 23, "date": 2020},
    {"age": 35, "date": 6320},
    {"age": 22, "date": 5020},
    {"age": 33, "date": 2320},
    {"age": 63, "date": 5024},
]


filtered_value = list(filter(lambda x: x["age"] == 23 or x["age"] == 33, students))
print(filtered_value)
```

```python
# mapping
nums = [1, 2, 3, 4, 5, 6, 8, 4, 3, 45]


def check_even(n):
    if n % 2 == 0:
        return {n: "Even"}
    return {n: "Odd"}


mapped_value = list(map(check_even, nums))
print(mapped_value)
```
