```python
# using sql alchemy
import sqlalchemy


from sqlalchemy import create_engine


engine = create_engine("sqlite:///mydb.db", echo=True)
```

```python
from sqlalchemy.ext.declarative import declarative_base
from sqlalchemy import Column, String, Integer


Base = declarative_base()


class Student(Base):
    __tablename__ = "students"
    id = Column(Integer, primary_key=True)
    name = Column(String)
    age = Column(Integer)


Base.metadata.create_all(engine)
```

```python
from sqlalchemy.orm import sessionmaker


Session = sessionmaker(bind=engine)
session = Session()
```

```python
student1 = Student(name="Himal", age=12)
student2 = Student(name="Sajina", age=12)


session.add(student1)
session.add(student2)
session.commit()
```

```python
# querying data from the sqlalchemy


students = session.query(Student).all()
print(students)
for student in students:
    print(student.id, student.name, student.age)
```

```python
import pandas as pd

data = {
    "City": ["Kathmandu", "Dharan", "Jhapa"],
    "Age": [20, 34, 35],
}


df1 = pd.DataFrame(data)
print(df1)
```

```python
data2 = {
    "City": ["Kathmandu", "Binaslu", "Gooday"],
    "Marks": [20, 34, 35],
}


# inner join vaneko intersection ho
# outer join vaneko chai union ho
# left join vaneko ko aadhar ma haalxa.
# right join vaneko right ko aadhar ma hunxa.


df2 = pd.DataFrame(data2)
df2
```

```python
merged_one = pd.merge(df1, df2, on="City", how="inner")
print(merged_one)
```

```python
union_merged = pd.merge(df1, df2, on="City", how="outer")
print(union_merged)


union_merged.fillna(12)
```

```python
left_join = pd.merge(df1, df2, on="City", how="left")
left_join
```

```python
right_join = pd.merge(df1, df2, on="City", how="right")
print(right_join)
```
