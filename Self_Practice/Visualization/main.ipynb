```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
import pandas as pd
```

```python
X = np.linspace(1,20,20)
y = 1.234 * np.square(X) + 2.345

X
```

```python
plt.figure(figsize=(10, 4))
plt.plot(X, y)
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Basic Matplotlib")
plt.show()
```

```python
fig = plt.figure(figsize=(10, 8))
type(fig)

# so figure() yesle aauta blank page jasto dinxa. (plt.plot() le chai sabai akkei choti garxa. but sub_plot le chai object oriented approach use garxa.)

ax1 = fig.add_subplot(2, 2, 1)
ax1.plot(X, y, c="r")

ax2 = fig.add_subplot(2, 2, 2)
ax2.plot(X, y, c="green")

ax3 = fig.add_subplot(2, 2, 3)
ax3.plot(X, y, c="blue")

ax4 = fig.add_subplot(2, 2, 4)
ax4.plot(X, y, c="b")

plt.tight_layout()
```

```python
movies = ["Krish3", "Mohenjodaro", "Saaho"]
quantity = [1000, 2000, 2334]

fig, axes = plt.subplots(2, 1)
print(fig)
print(axes)


axes[0].plot(X, y, c="r")
axes[0].set_title("Curve")
axes[0].set_xlabel("X")
axes[0].set_ylabel("Y")

axes[1].bar(movies, quantity, color=["Red", "Green", "Yellow"])
axes[1].set_title("Movies and Audience")
axes[1].set_xticks(ticks=movies)
axes[1].set_xticklabels(labels=movies, rotation=25)
axes[1].set_ylabel("Y")

plt.tight_layout()
```

```python
random_data = np.random.normal(loc=5, scale=10, size=1000)
print(random_data)
plt.hist(random_data, bins=60, color="gray", edgecolor="black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```
