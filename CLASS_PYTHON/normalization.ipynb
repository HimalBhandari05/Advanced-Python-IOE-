```python
# Normalization Concept.

import numpy as np
from sklearn.preprocessing import MinMaxScaler


X = np.array([[10], [20], [30], [40], [100]])
print(X)
scalar = MinMaxScaler(feature_range=(0, 1))
X_transform = scalar.fit_transform(X)
print(X_transform)
```

```python
# Z score normalization.

from sklearn.preprocessing import StandardScaler

X = np.array([[10], [20], [30], [40], [100]])
scalarr = StandardScaler()
x_transformed = scalarr.fit_transform(X)
print(x_transformed)
```

```python
# Normalization by decimal scaling.

X = np.array([[10], [20], [30], [40], [100], [200], [454]])

j = len(str(np.max(np.abs(X))))
x_transformedd = X / 10**j


print(x_transformedd)
```

```python
# Feature endoding.

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
cities = ["Kathmandu", "Pokhara", "Biratnagar"]
encoded = le.fit_transform(cities)
print(encoded)
```

```python
# One hot encoding.
import pandas as pd


df = pd.DataFrame({"source": ["river", "lake", "well"]})

df_encoded = pd.get_dummies(df, columns=["source"])
print(df_encoded)
```

```python
# One hot encoding using sklearn
from sklearn.preprocessing import OneHotEncoder

oee = OneHotEncoder()
objD = [["male"], ["female"], ["male"], ["female"]]
encoded_onehot = oee.fit_transform(objD)
print(encoded_onehot)
```

```python
# Ordinal encoding


from sklearn.preprocessing import OrdinalEncoder


encoder = OrdinalEncoder(categories=[["poor", "fair", "good"]])
quality = [['poor'],['good'],['fair']]
encoded = encoder.fit_transform(quality)
print(encoded)
```
