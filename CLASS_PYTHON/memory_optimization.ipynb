```python
"""
Memory optimize garna -> use smaller data types.
"""



import numpy as np
import time

start = time.time()

X = np.random.random(10000000)
for i in range(X.shape[0]):
    X[i] = X[i] + 2

end = time.time()

print(X[:10])

print(f"The time taken is {end-start} seconds")
```
