```python
import csv

with open(file="file.csv", mode="r") as f:
    content = csv.reader(f)
    print(content)
    for cont in content:
        print(cont)
```

```python
import csv

with open(file="file.csv", mode="r") as f:
    content = csv.DictReader(f)
    print(content)
    for cont in content:
        print(cont) 
```

```python
data = {'Name': 'Himjin', 'Address': 'Gauriganj', 'Age': '32'}

with open(file="file.csv" ,mode="a") as f:
    writer = csv.DictWriter(f, fieldnames=['Name', "Address" , "Age"])
    writer.writeheader()
    writer.writerow(data)
```
