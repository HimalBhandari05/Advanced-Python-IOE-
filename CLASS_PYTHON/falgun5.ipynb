```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
X = np.arange(1, 20, 0.27)
y = 1.123 * np.square(X) + 2.456

plt.figure(figsize=(6, 4))
plt.plot(X, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("X vs Y")
plt.grid()
plt.show()
```

```python
fig = plt.figure()

ax1 = fig.add_subplot(2, 2, 1)
ax1.plot(X, y, c="r")
ax1.set_xlabel("X-axis")
ax1.set_ylabel("Y-axis")
ax1.set_title("X vs Y")


ax2 = fig.add_subplot(2, 2, 2)
ax2.plot(X, y, c="g")
ax2.set_xlabel("X-axis")
ax2.set_ylabel("Y-axis")
ax2.set_title("X vs Y")

plt.tight_layout()
```

```python
fruits = ["Apples", "Bananas", "Cucumber"]
quantities = [100, 200, 80]
fig, axes = plt.subplots(2, 2)
print(axes)


for i in range(2):
    for j in range(2):
        axes[i, j].bar(fruits, quantities, color=["Red", "Yellow", "Green"], alpha=0.9)
        axes[i, j].set_title(f"Plot {i},{j}")
        axes[i, j].set_xticks(ticks=fruits)
        axes[i, j].set_xticklabels(labels=fruits, rotation=45)

plt.tight_layout()

# fig.subplots_adjust(wspace=1,hspace=1)
```

```python
data = np.random.normal(loc=50, scale=10, size=1000)

plt.hist(data, bins=20, color="skyblue", edgecolor="black")


plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Random data")

plt.show()
```

```python
from sklearn.datasets import make_moons

X, y = make_moons(n_samples=10000, noise=0.03, random_state=42)
print(X[:3], y[:3])


fig, axes = plt.subplots()
axes.scatter(X[:, 0], X[:, 1], c=y)
axes.text(x=1.3, y=1, s="Hello World", family="monospace")
axes.set_title("Title hai.")
plt.savefig("scatter.pdf")
```

```python
from sklearn.datasets import load_iris

X, y = load_iris(return_X_y=True, as_frame=True)
print(X.head())
```

```python
fig, axes = plt.subplots(2, 2)
axes[0, 0].bar(
    x=["Setosa", "Versicolor", "Virginica"],
    height=y.value_counts(),
    color=["Red", "Green", "Blue"],
)
axes[0, 0].set_title("Species Distribution")


axes[0, 1].scatter(X["sepal length (cm)"], X["petal length (cm)"], c=y)
axes[0, 1].set_title("Scatter plot sepal vs petal length")


axes[1, 0].hist(X["sepal length (cm)"], bins=10, edgecolor="black", color="skyblue")
axes[1, 0].set_title("Histogram of sepal length")


axes[1, 1].boxplot(X["petal length (cm)"])
axes[1, 1].set_title("Box plot")
```
