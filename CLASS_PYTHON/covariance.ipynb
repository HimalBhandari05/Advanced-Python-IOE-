```python
import numpy as np 
import pandas as pd


x = np.array([100,200,300,400])
y = np.array([12,34,54,64])

# matra duita value change hunxa vanne vanxa. we can't compare relationship across variables
np.cov(x,y)
```

```python
# Correlation -> standardized version of covariance. It is a measure of linear relationship between two variables. It ranges from -1 to 1.
# Correlation is unitless and does not depend on the scale of the variables. It is a measure of the strength and direction of the linear relationship between two variables. It is calculated as the
# Correlation = Covariance / (Standard Deviation of x * Standard Deviation of y)

np.corrcoef(x,y)
```

```python
import seaborn as sns
from sklearn.datasets import load_iris
import pandas as pd


X,Y = load_iris(as_frame=True, return_X_y=True)
X.corr()
```

```python
sns.heatmap(X.corr(), annot=True)
```

```python
# Skewness. left tira lean ki right tira lean vako kura dinxa. ( mean , median , mode)

x = np.array([10,230,34,453,4545])

y = np.array([10,20,34,43,345])


df = pd.DataFrame({
    "X":x,
    "Y":y
})
print(df)

print("Skewness of X column", df['X'].skew())
print("Skewness of Y column", df['Y'].skew())
```

```python
# Why kurtosis.


x = np.array([10,230,34,453,4545])

y = np.array([10,20,34,43,345])


df = pd.DataFrame({
    "X":x
})
print(df)

print("Kurtosis is:" , df['X'].kurt())
```
